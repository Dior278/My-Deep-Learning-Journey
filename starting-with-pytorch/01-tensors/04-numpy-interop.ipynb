# NumPy and PyTorch

Two conversions matter:

| Method | Copies? | Notes |
| --- | --- | --- |
| `torch.tensor(np_array)` | yes | Safe default |
| `torch.from_numpy(np_array)` | **no** | Same memory. Changing one changes the other |
| `tensor.numpy()` | **no** | Tensor must be on **CPU** and not require grad |
| `tensor.clone().numpy()` | yes | Isolated copy |

`.detach()` also shares storage. We use `.clone()` when we need independence.


````python
import torch
import numpy as np
````

## 1. Shared memory with `from_numpy`


````python
numpy_array = np.array([[1, 2], [3, 4]], dtype=np.float32)
pytorch_tensor = torch.from_numpy(numpy_array)

numpy_array[0, 0] = 99
print("tensor after editing NumPy:\n", pytorch_tensor)

pytorch_tensor[1, 1] = -1
print("NumPy after editing tensor:\n", numpy_array)
````

## 2. Isolated copy


````python
another_tensor = torch.tensor([[5.0, 6.0], [9.0, 8.0]])
numpy_from_tensor = another_tensor.clone().numpy()

another_tensor[1, 0] = 7.0
print("clone is unchanged:", numpy_from_tensor[1, 0])  # 9.0
````

## 3. `tensor.numpy()` on CPU

If the tensor is on GPU, call `.cpu()` first:

```python
cpu_from_gpu = gpu_tensor.cpu().numpy()
```


````python
cpu_tensor = torch.tensor([[10.0, 20.0], [30.0, 40.0]])
numpy_array_converted = cpu_tensor.numpy()

cpu_tensor[0, 1] = 25.0
print("NumPy sees the in-place edit:\n", numpy_array_converted)

numpy_array_converted[1, 0] = 35.0
print("tensor sees the NumPy edit:\n", cpu_tensor)
````
