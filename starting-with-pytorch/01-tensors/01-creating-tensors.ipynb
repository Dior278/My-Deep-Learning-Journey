# Creating tensors

A **tensor** is a multi-dimensional array. In PyTorch it can live on CPU or GPU, hold a dtype (`float32`, `int64`, …), and optionally track gradients.

Ways we create tensors.


## 1. Device and imports


```python
import torch
import numpy as np

print(f"PyTorch version: {torch.__version__}")

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
if device.type == "cuda":
    print(f"CUDA is available. GPU: {torch.cuda.get_device_name(0)}")
else:
    print("CUDA not available. Using CPU.")

print(f"Default device: {device}")
```

## 2. From Python lists

`torch.tensor(...)` infers the dtype (integers become `int64`, floats become `float32` unless we specify otherwise).


```python
data_list = [[1, 2], [3, 4]]
tensor_from_list = torch.tensor(data_list)

print(tensor_from_list)
print("dtype:", tensor_from_list.dtype)
print("shape:", tensor_from_list.shape)

float_tensor = torch.tensor(data_list, dtype=torch.float32)
print("forced float32:", float_tensor.dtype)
```

## 3. From NumPy

`torch.tensor(numpy_array)` **copies** the data. Later we will see `torch.from_numpy`, which **shares** memory.


```python
data_numpy = np.array([[5.0, 6.0], [7.0, 8.0]])
tensor_from_numpy = torch.tensor(data_numpy)

print(tensor_from_numpy)
print("dtype:", tensor_from_numpy.dtype)
print("shape:", tensor_from_numpy.shape)
```

## 4. By shape: zeros, ones, empty

`torch.empty` allocates memory without initializing it. Values are garbage until we write to them — we do not use it unless we overwrite every entry.


```python
shape = (2, 3)

zeros_tensor = torch.zeros(shape, dtype=torch.float64)
ones_tensor = torch.ones(shape)
empty_tensor = torch.empty(shape)

print("zeros:\n", zeros_tensor)
print("ones:\n", ones_tensor)
print("empty (uninitialized):\n", empty_tensor)

range_tensor = torch.arange(start=0, end=10, step=2)
print("arange:", range_tensor)
```

## 5. Random tensors

- `torch.rand`: uniform on `[0, 1)`
- `torch.randn`: standard normal (mean 0, std 1)


```python
torch.manual_seed(0)

rand_tensor = torch.rand(shape)
randn_tensor = torch.randn(shape)

print("uniform:\n", rand_tensor)
print("normal:\n", randn_tensor)
```

## 6. Same shape as another tensor (`*_like`)

These helpers copy **shape and dtype** (unless we override them).


```python
first = torch.zeros_like(rand_tensor)
second = torch.ones_like(randn_tensor)
third = torch.randn_like(randn_tensor)
fourth = torch.rand_like(zeros_tensor)

print(first)
print(second)
print(third)
print(fourth)
```
